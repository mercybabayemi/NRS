```python
transactions['transaction_date'] = pd.to_datetime(transactions['transaction_date'])
```

```python
transactions['amount'] = transactions['amount'].replace({',': ''}, regex=True)

transactions['amount'] = pd.to_numeric(transactions['amount'], errors='coerce')
```

```python
individual_accounts['created_at'] = pd.to_datetime(individual_accounts['created_at'])
```

```python
individuals_outflows = transactions.groupby('from_account')['amount'].agg(['sum', 'count']).reset_index()
```

```python
individual_account_list = individuals_economic_activities['account_number'].unique()
individual_account_set = set(individual_account_list)
```

```python
expense_detail = transactions[transactions['from_account'].isin(individual_account_set)].copy()
```

```python
expense_detail['transaction_date'] = pd.to_datetime(expense_detail['transaction_date'])
expense_detail['quarter'] = expense_detail['transaction_date'].dt.to_period('Q')
expense_detail['month'] = expense_detail['transaction_date'].dt.strftime('%B')
```

```python
expense_detail['spend_category'] = expense_detail['transaction_type'].fillna(expense_detail['narration'])
```

```python
expense_analysis_df = expense_detail.merge(
    individuals_economic_activities[['account_number', 'full_name', 'individual_id']], 
    left_on='from_account', 
    right_on='account_number', 
    how='left'
)
```

```python
print(expense_analysis_df[['full_name', 'amount', 'spend_category', 'transaction_date', 'quarter']].head())
```

```python
expense_analysis_df.head()
```

```python

master_spending_pivot = pd.pivot_table(
    expense_analysis_df, 
    index=['individual_id'], 
    columns='spend_category', 
    values='amount', 
    aggfunc=['sum', 'count'], 
    fill_value=0
)
```

```python
display(master_spending_pivot.head(10))
```

```python
import matplotlib.pyplot as plt

# 1. Choose the person (using the ID you want to analyze)
target_id = 1 # Replace with an actual ID from your table

# 2. Extract the row for that person
person_row = master_spending_pivot.loc[target_id]

# 3. Get the list of all categories (Business_purchase, gift, etc.)
# We grab the names from the 'sum' level
labels = master_spending_pivot['sum'].columns.tolist()

# 4. Get the values for ALL categories
# These are arrays of numbers (e.g., [5000, 200, 1500...])
sums = person_row['sum'].values.flatten()
counts = person_row['count'].values.flatten()

# 5. Build the Chart
fig, ax1 = plt.subplots(figsize=(12, 6))

# The Blue Bars (Total Money)
ax1.bar(labels, sums, color='skyblue', alpha=0.7, label='Total Spent')
ax1.set_ylabel('Total Amount Spent ($)', color='blue', fontsize=12)

# The Red Line (Frequency)
ax2 = ax1.twinx()
ax2.plot(labels, counts, color='red', marker='o', linewidth=2, label='Transaction Count')
ax2.set_ylabel('Number of Transactions', color='red', fontsize=12)

plt.title(f"Spending Velocity Profile: ID {target_id}", fontsize=14)
ax1.set_xticklabels(labels, rotation=45) # Tilt the names so they don't overlap
fig.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

target_id = 2
# 1. Select the individual row
# Ensure target_id is the index value (e.g., an integer)
person_row = master_spending_pivot.loc[target_id]

# 2. Extract and Flatten the values
# .values returns the numbers, .flatten() makes them a simple list [5000, 200, ...]
labels = master_spending_pivot['sum'].columns.tolist()
sums = person_row['sum'].values.flatten()
counts = person_row['count'].values.flatten()

# 3. Create the Plot
fig, ax1 = plt.subplots(figsize=(12, 6))

# --- LEFT SIDE: THE MONEY (BARS) ---
ax1.bar(labels, sums, color='skyblue', alpha=0.7, label='Total Value')
ax1.set_ylabel('Total Amount Spent (Currency)', color='blue', fontsize=12, fontweight='bold')
ax1.tick_params(axis='y', labelcolor='blue')

# Format the money axis with commas (e.g., 5,000)
ax1.yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))

# --- RIGHT SIDE: THE FREQUENCY (LINE) ---
ax2 = ax1.twinx()
ax2.plot(labels, counts, color='red', marker='o', linewidth=2, markersize=8, label='Frequency')
ax2.set_ylabel('Number of Transactions (Frequency)', color='red', fontsize=12, fontweight='bold')
ax2.tick_params(axis='y', labelcolor='red')

# Ensure the frequency axis only shows whole numbers (no 1.5 transactions)
ax2.yaxis.set_major_locator(ticker.MaxNLocator(integer=True))

# 4. Final Formatting
plt.title(f"Spending Velocity Analysis: ID {target_id}", fontsize=15, pad=20)
ax1.set_xticklabels(labels, rotation=45, ha='right')

# Add a grid for easier reading of the bars
ax1.grid(axis='y', linestyle='--', alpha=0.3)

fig.tight_layout()
plt.show()
```

```python
# 1. Group by recipient to see who is getting the most money
hubs = transactions.groupby('to_name').agg({
    'amount': 'sum',           # Total value received
    'transaction_id': 'count', # Total frequency (volume)
    'from_account': 'nunique'  # How many different people sent them money
}).reset_index()

# 2. Identify Hubs: High nunique 'from_account' means many people depend on them
hubs.columns = ['Recipient', 'Total_Received', 'Frequency', 'Unique_Senders']
economic_hubs = hubs.nlargest(10, 'Unique_Senders')

print("Top 10 Economic Hubs:")
display(economic_hubs)
```

```python
economic_hubs.sort_values('Unique_Senders').plot(
    x='Recipient', 
    y='Unique_Senders', 
    kind='barh', 
    color='skyblue', 
    figsize=(10, 6),
    legend=False
)
plt.title('Hub Reach: Number of Unique Senders per Recipient')
plt.xlabel('Count of Unique Individuals')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# 1. Prepare your data from the hub analysis we did earlier
# We sort by 'Unique_Senders' to show the most influential hubs first
hubs_sorted = economic_hubs.sort_values('Unique_Senders', ascending=False)

labels = hubs_sorted['Recipient'].tolist()
reach = hubs_sorted['Unique_Senders'].tolist()
value = hubs_sorted['Total_Received'].tolist()

# 2. Create the Figure and the First Axis (Bars for Value)
fig, ax1 = plt.subplots(figsize=(14, 7))

# Plot the BARS (The "Value" of the Hub)
ax1.bar(labels, value, color='skyblue', alpha=0.7, label='Total Value Received')
ax1.set_ylabel('Total Value (Currency)', color='blue', fontsize=12, fontweight='bold')
ax1.tick_params(axis='y', labelcolor='blue')

# Format the left axis to show currency with commas (e.g., 5,000)
ax1.yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))

# 3. Create the Second Axis (Line for Reach)
# This axis shares the same X-axis as the bars
ax2 = ax1.twinx()

# Plot the LINE (The "Reach" of the Hub)
ax2.plot(labels, reach, color='red', marker='o', linewidth=3, markersize=10, label='Unique Senders')
ax2.set_ylabel('Reach (Unique Senders)', color='red', fontsize=12, fontweight='bold')
ax2.tick_params(axis='y', labelcolor='red')

# 4. Final Formatting
plt.title('2026 Economic Hubs: Community Reach vs. Financial Value', fontsize=16, pad=20)
ax1.set_xticklabels(labels, rotation=45, ha='right')

# Add a light grid to help read the bars
ax1.grid(axis='y', linestyle='--', alpha=0.3)

fig.tight_layout()
plt.show()
```

```python
likely_relationships = (
    transactions
    .groupby(['from_account', 'to_account'], as_index=False)
    .agg(
        total_amount=('amount', 'sum'),
        transaction_count=('amount', 'count')
    )
    .query("transaction_count >= 2 and total_amount >= 50000")
)
```

```python
# 1. Find people who pay the most unique individuals
employer_check = likely_relationships.groupby('from_account')['to_account'].nunique().reset_index()
employer_check.columns = ['Sender', 'Number_of_Recipients']

# 2. Sort and show the top 10 "Potential Employers"
top_employers = employer_check.nlargest(10, 'Number_of_Recipients')

# 3. Plot
plt.figure(figsize=(10, 6))
plt.bar(top_employers['Sender'], top_employers['Number_of_Recipients'], color='orange')

plt.title("Potential Employers: Who is paying the most people?", fontsize=14)
plt.ylabel("Count of Unique Recipients")
plt.xlabel("Sender Name")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Create a "Relationship" label for the X-axis
likely_relationships['connection'] = likely_relationships['sender_name'] + " ➔ " + likely_relationships['recipient_name']

# 2. Sort by the total amount to find the biggest money flows
top_10_flows = likely_relationships.nlargest(10, 'total_amount')

# 3. Plot the Bar Chart
plt.figure(figsize=(12, 6))
plt.bar(top_10_flows['connection'], top_10_flows['total_amount'], color='teal')

plt.xticks(rotation=45, ha='right')
plt.title("Top 10 Person-to-Person Money Flows (Jan 2026)", fontsize=14)
plt.ylabel("Total Amount Transferred")
plt.xlabel("Sender ➔ Recipient")
plt.tight_layout()
plt.show()
```
